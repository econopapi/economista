# Crecimiento económico con IMF DataMapper/WEO

Este ejemplo usa la API pública IMF DataMapper API v2 para descubrir y descargar crecimiento real del PIB desde el World Economic Outlook (WEO). DataMapper es una superficie anual de IMF; no es el API SDMX completo.

```python
from economista import data
```

## Descubrir el indicador

La búsqueda se limita a IMF, al catálogo de indicadores y a WEO. El catálogo informa unidad, actualización y año de proyección de la serie.

```python
indicators = data.search(
    source="imf",
    dataset="WEO",
    query="real GDP growth",
    kind="indicators",
)

indicators[["id", "name", "unit", "projection_year", "last_modified"]]
```

## Descargar crecimiento real del PIB

`NGDP_RPCH` mide el cambio porcentual anual del PIB real. Consultamos México y Colombia para obtener un `EconDataset` en el mismo esquema canónico que las demás fuentes.

```python
growth = data.fetch(
    source="imf",
    dataset="WEO",
    indicator="NGDP_RPCH",
    geo=["MEX", "COL"],
    start=2020,
    end=2026,
)

growth_frame = growth.to_pandas()
growth_frame[["geo", "geo_name", "time", "value", "unit", "dataset"]]
```

## Comparar países

El valor ya está expresado como porcentaje anual, por lo que basta pivotar la tabla larga para comparar las economías. Revisa `projection_year` arriba antes de interpretar períodos futuros como observaciones realizadas.

```python
growth_percent = (
    growth_frame.pivot(index="time", columns="geo", values="value")
    .sort_index()
)

growth_percent.round(1)
```
